```python
# In a notebook cell FIRST (before importing pyplot)
%matplotlib inline
# or: %matplotlib widget  (interactive pan/zoom, if ipympl is installed)

from empirical_bench_data.empirical_bench import load, gen_dataset
import matplotlib.pyplot as plt

data = load()
keys = [p["key"] for p in data["problems"].values()]

for key in keys:
    X, y = gen_dataset(data, key, scale_dataset=True)

    x0 = X.iloc[:, 0].values
    plt.figure()
    plt.scatter(x0, y, s=8)
    plt.xlabel(str(X.columns[0]))
    plt.ylabel("y")
    plt.title(key)
    plt.tight_layout()
    plt.show()
```
