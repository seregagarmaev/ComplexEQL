# Description

In this notebook, we apply PySR algorithm to symbolic regression benchmarks.

```python
X = np.random.randn(300, 30)
y = X[:, 3]**2 - X[:, 19]**2 + 1.5
```

```python
model = PySRRegressor(
    binary_operators=["+", "-", "*", "/"],
    unary_operators=["exp"],
    select_k_features=5,
)
```

```python
model.fit(X, y)
```
